# Exploratory Analysis -- AI Email Spam Classifier

Quick, standalone exploration of the SMS Spam Collection dataset.
This notebook is for exploration only; the real, reproducible training
pipeline lives in `src/train_model.py`.

````
import sys, os
sys.path.insert(0, os.path.abspath(".."))

from src.data_loader import load_and_clean_dataset
from src.preprocessing import clean_series

df, stats = load_and_clean_dataset(os.path.join("..", "data", "spam.csv"))
print(stats)
df.head()
````

````
import matplotlib.pyplot as plt

df["label"].map({0: "Ham", 1: "Spam"}).value_counts().plot(kind="bar", color=["#2ecc71", "#e74c3c"])
plt.title("Class Distribution")
plt.show()
````

````
df["length"] = df["message"].str.len()
df.groupby("label")["length"].describe()
````

````
df["cleaned_message"] = clean_series(df["message"])
df[["message", "cleaned_message"]].sample(5, random_state=1)
````
